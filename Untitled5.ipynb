```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style for plots
sns.set_theme(style="whitegrid")

# If you uploaded your file manually or via Google Drive, update the filename below:
file_path = 'ecommerce_customer_data_large.csv' # or 'ecommerce_customer_data_custom_ratios.csv'
df = pd.read_csv(file_path)

# Display basic info
print("Dataset Shape:", df.shape)
df.head()
```

```python
# 1. Drop duplicate columns if any (e.g., duplicate Age columns)
if 'Age' in df.columns and 'Customer Age' in df.columns:
    df = df.drop(columns=['Age'])

# 2. Handle Missing Values
# Fill missing Returns with 0 (assuming blank means no return)
if 'Returns' in df.columns:
    df['Returns'] = df['Returns'].fillna(0)

# 3. Fix Data Types
df['Purchase Date'] = pd.to_datetime(df['Purchase Date'])

# 4. Feature Engineering: Extract date parts
df['Purchase_Year'] = df['Purchase Date'].dt.year
df['Purchase_Month'] = df['Purchase Date'].dt.month
df['Day_Of_Week'] = df['Purchase Date'].dt.day_name()

print("Data Cleaning Complete. Missing values handled.")
```

```python
# Calculate Reference Date (1 day after max purchase date in dataset)
snapshot_date = df['Purchase Date'].max() + pd.Timedelta(days=1)

# Group by Customer ID to calculate RFM metrics
rfm = df.groupby('Customer ID').agg({
    'Purchase Date': lambda x: (snapshot_date - x.max()).days, # Recency
    'Customer ID': 'count',                                    # Frequency
    'Total Purchase Amount': 'sum'                             # Monetary
})

rfm.columns = ['Recency', 'Frequency', 'Monetary']

# Assign RFM Quartile Scores (1 to 4, where 4 is the best)
rfm['R_Score'] = pd.qcut(rfm['Recency'], 4, labels=[4, 3, 2, 1]) # Lower recency is better
rfm['F_Score'] = pd.qcut(rfm['Frequency'].rank(method='first'), 4, labels=[1, 2, 3, 4])
rfm['M_Score'] = pd.qcut(rfm['Monetary'], 4, labels=[1, 2, 3, 4])

# Combine scores to create an RFM Segment string
rfm['RFM_Score'] = rfm['R_Score'].astype(str) + rfm['F_Score'].astype(str) + rfm['M_Score'].astype(str)

# Simple customer tier labeling
def segment_customer(df_rfm):
    if df_rfm['M_Score'] >= 3 and df_rfm['F_Score'] >= 3:
        return 'Champions / High Value'
    elif df_rfm['R_Score'] <= 2 and df_rfm['F_Score'] >= 3:
        return 'At Risk / Churning'
    else:
        return 'Regular / Low Engagement'

rfm['Customer_Segment'] = rfm.apply(segment_customer, axis=1)
print(rfm['Customer_Segment'].value_counts())
```

```python
# 1. Purchase Distribution by Product Category
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='Product Category', order=df['Product Category'].value_counts().index, palette='viridis')
plt.title('Transactions by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Count')
plt.show()

# 2. Monthly Sales Trend Over Time
plt.figure(figsize=(12, 5))
monthly_sales = df.groupby(df['Purchase Date'].dt.to_period('M'))['Total Purchase Amount'].sum()
monthly_sales.plot(kind='line', marker='o', color='b')
plt.title('Total Sales Trend Over Time')
plt.xlabel('Month-Year')
plt.ylabel('Total Purchase Amount')
plt.grid(True)
plt.show()
```

```python
# 6. Exporting Results & Generating Final Summary Report DataFrame
# Create a summary dataframe for reporting
segment_summary = rfm.groupby('Customer_Segment').agg(
    Customer_Count=('Recency', 'count'),
    Avg_Recency=('Recency', 'mean'),
    Avg_Frequency=('Frequency', 'mean'),
    Total_Monetary=('Monetary', 'sum')
).reset_index()

print("\n--- Segment Summary Table ---")
display(segment_summary)

# Save processed RFM results to CSV for your final project submission
rfm.to_csv('customer_rfm_segments.csv')
print("\nSuccess! RFM segmentation results have been saved to 'customer_rfm_segments.csv' for your project report.")
```
